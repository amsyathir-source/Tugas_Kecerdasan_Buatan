# Optimasi Jumlah Geophone untuk Survei Seismik Dangkal

## 1. Latar Belakang

Dalam survei seismik dangkal, jumlah geophone dapat memengaruhi kualitas data dan biaya akuisisi. Penambahan geophone umumnya membantu meningkatkan cakupan pengukuran, tetapi juga menambah kebutuhan peralatan dan waktu pemasangan. Oleh karena itu, diperlukan pendekatan optimasi untuk menyeimbangkan kualitas data dan biaya.

**Catatan:** angka pada contoh ini merupakan data sintetis/ilustratif untuk demonstrasi metode, bukan hasil pengukuran lapangan.

## 2. Tujuan

Menentukan jumlah geophone yang memberikan kompromi antara nilai kesalahan (RMSE) yang rendah dan biaya relatif yang masih terkendali.

## 3. Struktur Prompt (R–T–C–I–K–O–Q–V)

- **Role:** Asisten analisis data geofisika.
- **Task:** Mengevaluasi beberapa pilihan jumlah geophone berdasarkan RMSE dan biaya relatif.
- **Context:** Survei seismik dangkal dengan jumlah geophone yang berbeda.
- **Input:** Jumlah geophone, RMSE ilustratif, dan model biaya relatif.
- **Constraints:** Data hanya untuk contoh; hasil tidak boleh dianggap sebagai rekomendasi lapangan tanpa validasi.
- **Output:** Tabel perbandingan, grafik, dan pilihan berdasarkan fungsi objektif.
- **Quality criteria:** Perhitungan konsisten, grafik terbaca, serta asumsi dijelaskan.
- **Verification:** Periksa kembali rumus, data masukan, dan indeks nilai objektif minimum.

## 4. Data dan Metode

Jumlah geophone yang diuji adalah 4, 6, 8, 10, 12, 14, 16, 18, dan 20. RMSE merupakan ukuran kesalahan; semakin kecil RMSE, semakin kecil kesalahan pada contoh ini.

Biaya relatif dimodelkan secara sederhana:

\[
C = 1.2N
\]

dengan \(N\) adalah jumlah geophone. Untuk menyeimbangkan kualitas dan biaya, digunakan fungsi objektif:

\[
J = \frac{RMSE}{RMSE_{max}} + \lambda\frac{C}{C_{max}}
\]

dengan \(\lambda = 0.5\). Nilai \(J\) yang lebih kecil menunjukkan kompromi yang lebih baik berdasarkan bobot dan asumsi contoh ini.

**Pilihan terbaik menurut fungsi objektif contoh ini: 12 geophone**, dengan nilai objektif 0.662. Pilihan ini hanya optimal untuk data sintetis, rumus biaya, dan bobot yang digunakan di sini.

## 7. Kode Python

Kode berikut dapat dijalankan di Python/Jupyter untuk menghasilkan grafik dan mencari nilai objektif minimum.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

jumlah = np.array([4, 6, 8, 10, 12, 14, 16, 18, 20])
rmse = np.array([18.5, 13.2, 10.1, 8.0, 6.7, 5.9, 5.4, 5.1, 4.9])
biaya = 1.2 * jumlah

lambda_biaya = 0.5
objektif = rmse / rmse.max() + lambda_biaya * biaya / biaya.max()

hasil = pd.DataFrame({
    "Jumlah geophone": jumlah,
    "RMSE (m)": rmse,
    "Biaya relatif": biaya,
    "Nilai objektif": objektif
})
print(hasil.round(3))
print("Jumlah geophone terbaik:", jumlah[np.argmin(objektif)])

fig, ax1 = plt.subplots(figsize=(9, 5))
ax1.plot(jumlah, rmse, marker="o", label="RMSE (m)")
ax1.set_xlabel("Jumlah geophone")
ax1.set_ylabel("RMSE (m)")
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.plot(jumlah, biaya, marker="s", linestyle="--", label="Biaya relatif")
ax2.set_ylabel("Biaya relatif")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right")
plt.title("Optimasi Jumlah Geophone untuk Survei Seismik Dangkal")
plt.tight_layout()
plt.show()
```

3. Model biaya \(C=1.2N\) adalah penyederhanaan. Biaya lapangan nyata dapat mencakup kabel, personel, waktu akuisisi, transportasi, dan kondisi lokasi.
